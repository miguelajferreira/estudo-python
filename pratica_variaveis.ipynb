Exercício 3: Escreve uma função chamada classificar_escore que receba um escore MoCA e retorne uma string indicando a categoria: "Normal" (escore ≥ 26), "Declínio Cognitivo Leve" (18 a 25) ou "Demência" (abaixo de 18). Testa a função com pelo menos três valores diferentes.

```python
def classificar_score(estado):
    if estado >= 26:
        print("Normal")
    elif 18 <= estado <= 25:
        print("Declínio Cognitivo Ligeiro")
    elif estado < 18:
        print("Demência")
    else:
        print("Something strange is happening")
    return estado

classificar_score(27)
```
